# Why one engine, and what this book will not cover

This book is for a student — undergraduate or graduate — learning statistics and Julia at the same
time. You may have used R a little. You have never opened Julia. Nobody expects you to have fitted
a mixed model, or to know what one is. Nothing on these pages assumes more than that: basic Julia
is Class 1, taught on the same data you model in Class 2, and every statistical word arrives at the
moment a model needs it. If that is you, you are the reader, and the climb is paced for you.

Here is the kind of question it answers. The house sparrows of Lundy Island have been caught,
measured, ringed and followed, nest by nest, for years, and the files from those years are the data
of this book. Does a bigger-boned bird carry a longer wing, and by how much? Why do some chicks die
in the nest while others fledge, and why does one female fledge more than another? When the same
bird turns up in the file four times, have you learned something four times, or once? And how much
of what a chick becomes did it get from its parents, and how much from the nest it grew up in?
Every one of those is a case of the sentence Class 2 opens on: *things vary, and some of the
variation has a reason.* A statistical model is the instrument for saying how much of the variation
has a reason, what the reason is, and how sure you are — and for being told, by the numbers, when
you have claimed too much. That is what this book teaches, on those birds, one question at a time:
the wing in Class 2, the chicks in Class 3, the bird measured four times in Class 6, the parents and
the nest in Class 10. The rungs between are what it takes to get from each of those questions to
the next one honestly.

It is a reboot of *Statistical Models with R: An Introduction with Sparrows*, drafted in 2009–2010 (the recovered files are dated November 2009 to March 2010) and
never published. What survives from that draft is the shape: an office hour rather than a lecture,
a small cast who each own an organism, real birds off a real island, and the habit of leaving the
mistakes in. What changed is the engine and the language, and the fact that every number in this version
comes from code that ran, not from a keyboard.

**The climb.** The book is not organised by topic. It is organised by **constants**. Each rung frees one thing
the rung before it held fixed: the shape of the noise, then the spread, then the correlation
between rows, then the slope, then the assumption that groups are independent of one another. The
diagram below draws the whole climb; it is the one picture worth looking at before you start.

**Two threads run the length of the climb**, and they are not the same kind of thing.

The first is Julia itself. Basic Julia is not an appendix and not a prerequisite — it is Class 1,
taught with the same data you model in Class 2.

The second is **simulation**. From Class 2 onward, most chapters end their Julia summary with one

thread in one place: from making random numbers repeatable, to checking whether an interval really
catches the truth as often as it claims, to re-fitting a model on data it generated itself.

**The exercises** are applied and dataset-driven, and every one of them asks you to explain your
code in your own words, to somebody who has not opened Julia. That instruction is the exercise.

**The data are real.** They are 171 house sparrows measured on Lundy Island for that course,

## What this book will not cover

**Meta-analysis and phylogenetic comparative methods.** Meta-analysis combines the results of
many published studies; phylogenetic comparative methods compare species while allowing for their
shared ancestry. Both are mentioned. Neither gets a
chapter, and the omission is deliberate rather than an oversight: they are the same statistical
problem wearing different clothes — non-independence that arrives when several published estimates come from the same study, and
non-independence that arrives from shared ancestry — and they belong together in a second book
rather than split across the back of this one.

Read that fence precisely. Class 10 *does* teach a model in which the groups are told how related
they are to one another, and a species tree (a phylogeny) is one of the four forms that idea takes. What is fenced off is the
field around it: the comparative question, the tree and its uncertainty, the effect-size
literature, the machinery for combining published estimates. You will meet the matrix. You will
not be taught the discipline that lives on it.

**Rungs eleven and twelve.** Version one of this book climbs to rung ten and stops. Modelling the
spread itself, and two responses at once with a correlation that varies, are held for a second
version. The coda says what they are and why they are the same argument continued rather than a
new one.

**Bayesian inference** — the approach where you state what you believed before seeing the data and
let the data update it. The engines do not do that; they work from the data alone. Every interval
in this book is a confidence interval, and every fit is found by maximum likelihood or its close
relative, REML; Class 8 is about exactly which of those two you got and were never asked to choose,
and each of those words is explained at the moment a model needs it, not here. That is a scope
decision, not a verdict on the alternative.

**Installation.** Class 1 teaches everything the language asks of you — files, data frames (tables
of data), pictures, and reading a wall of red text — and skips the install. Today the two engines
are installed straight from their GitHub pages, which takes a paragraph of instructions; it becomes
a single line the day they are published in the official package lists. Writing that paragraph now means writing it twice.

## Why one engine

It is a statistics book with one engine in it.

Not one language — there are two, R and Julia — but one *engine*: a single pair of twinned
packages, [`drmTMB`](https://github.com/itchyshin/drmTMB) in R and
[`DRM.jl`](https://github.com/itchyshin/DRM.jl) in Julia, that fit every model from the first
straight line to the last relatedness matrix — a table of how related every pair of animals is.
Nothing is swapped out halfway. When you reach the
chapter where rows stop being strangers, you do not also learn a new package, a new formula
grammar and a new summary layout on the same morning.

The climb is the reason. Each rung frees one thing, and if the software changes at the same moment
the model does, you cannot tell which of the two moved the number. A chapter that fits a random
intercept in a new package and reports a different standard error has taught you nothing, because
you cannot say whether the estimator changed, or only the printing did. Holding the engine still is
what makes the model the only moving part.

There is a second reason, less tidy and more honest. The climb needs models that can allow for
unequal effort or exposure per row, that let the scatter itself be modelled rather than fixed, that
can be told how the groups in the data are related to one another, and that end in a heritability —
all fitted the same way, from the data alone, in one frame. You will meet the name for each of
those at the moment a model needs it, not here. That combination is
awkward to assemble in Julia today out of the general-purpose packages, which is precisely why
these two engines exist. A book that stitched five packages together would spend its teaching time
on the stitching.

**Other packages appear in three boxes, and nowhere else.** A box has to earn its place, and the
only way it can is by stopping a misreading: two correct programs print different numbers for the
same model, and the box names the mechanism and gives you a boundary you can compute. There are
three in the whole book. Class 2's, where R's `lm` and this engine disagree about the residual
spread, and the reason is a divisor. Class 8's, where R's `lmer` and this engine disagree about a
variance, and the reason is that each chose a different estimator on your behalf and neither said
so. Class 9's, where two programs approximate the same integral by different rules, and the
disagreement is about neither program. Those are among the best pages in the book, and the reason
there are not more of them is the reason there are not more of anything here: writing every chapter
twice, in two languages, costs more attention than it returns — a finding about split attention
that is older than either language and does not care which two you pick (Chandler & Sweller 1991).

**Where this book sits.** It is not a tour of Julia or of its data ecosystem; there are good books
for that. It teaches you to fit a model for counts or yes/no outcomes measured on groups of related
individuals, allow for unequal effort, know exactly which method produced each number, and be able
to defend it.

## Nothing here is typed in

**Never invent output.** Every number, every printed table, every figure on every page of this book
is produced when the site is built, by running the code you can see above it. Nothing is typed from
memory, nothing is pasted from a session that is not shown, and when someone in the dialogue says a
number, that number was computed by the code on the page a moment earlier, not remembered and typed
in.

That is the rule that does most work here, because a chapter with plausible invented numbers is
worse than an obviously unfinished one: nobody catches it. So it is not a promise, it is a
mechanism: the page you are reading cannot exist unless its code ran. The smallest possible
demonstration of it is the colophon at the foot of this page, after the reading list, where the
only code on this page runs.

## Further reading

*Four, no more, because a preface is not a syllabus. The details of all four were checked on
2026-09-07 against OpenAlex, an open catalogue of research papers.*

1. **Bezanson, J., Edelman, A., Karpinski, S. & Shah, V. B. (2017) "Julia: A Fresh Approach to
   Numerical Computing", *SIAM Review* 59(1):65–98.** doi:10.1137/141000671. The language explained
   by the people who built it. Read it for one idea — that a function is a family of methods chosen
   by the types of *all* its arguments — because that idea is what makes an engine like this one
   possible to write, and it is also what the error messages in Class 1 are printing at you.
2. **Bolker, B. M., Brooks, M. E., Clark, C. J., Geange, S. W., Poulsen, J. R., Stevens, M. H. H. &
   White, J.-S. S. (2009) "Generalized linear mixed models: a practical guide for ecology and
   evolution", *Trends in Ecology & Evolution* 24(3):127–135.** doi:10.1016/j.tree.2008.10.008. The
   paper that showed biologists, not only statisticians, that two programs fitting the same model
   can print different numbers. Read it now for the map of where this book is going, and again after Class 8 for
   the disagreement it describes; its software advice has aged, its practice has not.
3. **Chandler, P. & Sweller, J. (1991) "Cognitive Load Theory and the Format of Instruction",
   *Cognition and Instruction* 8(4):293–332.** doi:10.1207/s1532690xci0804_2. The evidence behind
   the decision not to print every example twice. Splitting material a reader must mentally
   integrate across two places costs comprehension; that finding is why the R boxes in this book are
   rare and earned rather than routine.
4. **Peng, R. D. (2011) "Reproducible Research in Computational Science", *Science*
   334(6060):1226–1227.** doi:10.1126/science.1213847. Two pages on why code and data travelling
   with the claim is the minimum standard, not a virtue. It is the argument this book applies to
   itself, and the reason the rule above is a mechanism rather than a promise.
